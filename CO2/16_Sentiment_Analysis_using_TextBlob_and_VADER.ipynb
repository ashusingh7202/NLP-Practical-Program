```python
%pip install vaderSentiment

from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

text = "The product is excellent and I really love it!"

# TextBlob
polarity = TextBlob(text).sentiment.polarity
print("TextBlob polarity:", polarity)

if polarity > 0:
    print("TextBlob sentiment: Positive")
elif polarity < 0:
    print("TextBlob sentiment: Negative")
else:
    print("TextBlob sentiment: Neutral")

# VADER
analyzer = SentimentIntensityAnalyzer()
scores = analyzer.polarity_scores(text)
compound = scores["compound"]

print("\nVADER scores:", scores)

if compound >= 0.05:
    print("VADER sentiment: Positive")
elif compound <= -0.05:
    print("VADER sentiment: Negative")
else:
    print("VADER sentiment: Neutral")
```
