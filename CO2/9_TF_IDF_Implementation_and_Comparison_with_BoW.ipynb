```python

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

corpus = [
    "I love natural language processing",
    "I love machine learning",
    "Language processing is interesting"
]

bow = CountVectorizer()
tfidf = TfidfVectorizer()

print("Vocabulary:", bow.fit(corpus).get_feature_names_out())
print("\nBoW Matrix:\n", bow.transform(corpus).toarray())
print("\nTF-IDF Matrix:\n", tfidf.fit_transform(corpus).toarray())
```
