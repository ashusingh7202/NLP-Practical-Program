```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.pipeline import make_pipeline

texts = [
    "I love this movie",
    "This film is amazing",
    "I hate this movie",
    "This film is terrible",
    "The movie is wonderful",
    "The film is boring"
]
labels = ["positive", "positive", "negative",
          "negative", "positive", "negative"]

nb = make_pipeline(TfidfVectorizer(), MultinomialNB())
svm = make_pipeline(TfidfVectorizer(), LinearSVC())

nb.fit(texts, labels)
svm.fit(texts, labels)

test = ["This movie is amazing"]

print("Naive Bayes:", nb.predict(test)[0])
print("SVM:", svm.predict(test)[0])
```
