```python
%pip install POT

import gensim.downloader as api
from gensim.utils import simple_preprocess

model = api.load("glove-wiki-gigaword-50")

doc1 = simple_preprocess("I love machine learning")
doc2 = simple_preprocess("I enjoy artificial intelligence")

distance = model.wv if hasattr(model, "wv") else model

# Gensim KeyedVectors supports WMD with compatible installations
print("Document 1:", doc1)
print("Document 2:", doc2)
print("WMD:", distance.wmdistance(doc1, doc2))
```
