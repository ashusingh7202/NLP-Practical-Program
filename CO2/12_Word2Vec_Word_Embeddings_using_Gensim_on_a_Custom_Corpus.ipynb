```python
%pip install gensim

from gensim.models import Word2Vec

corpus = [
    ["machine", "learning", "is", "awesome"],
    ["deep", "learning", "is", "a", "subset", "of", "machine", "learning"],
    ["natural", "language", "processing", "uses", "word", "embeddings"]
]

model = Word2Vec(sentences=corpus, vector_size=50, window=3, min_count=1, workers=1)

print("Vector representation for 'learning':")
print(model.wv['learning'][:5]) # Display first 5 dimensions

print("\nMost similar words to 'learning':")
print(model.wv.most_similar('learning', topn=2))
```
