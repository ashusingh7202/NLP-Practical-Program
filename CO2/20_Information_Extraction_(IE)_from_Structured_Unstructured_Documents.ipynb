```python

import re

document = """
Contact: Rahul Sharma
Email: rahul@example.com
Phone: 9876543210
Meeting date: 15/10/2026
"""

emails = re.findall(
    r'[\w.-]+@[\w.-]+\.\w+', document
)

phones = re.findall(
    r'\b\d{10}\b', document
)

dates = re.findall(
    r'\b\d{2}/\d{2}/\d{4}\b', document
)

print("Extracted Information")
print("Emails:", emails)
print("Phone numbers:", phones)
print("Dates:", dates)
```
