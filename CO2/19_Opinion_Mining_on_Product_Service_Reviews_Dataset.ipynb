```python

from textblob import TextBlob

reviews = [
    "This phone has an excellent camera",
    "Battery life is very poor",
    "The product is average",
    "I am extremely happy with this purchase",
    "The quality is terrible"
]

for review in reviews:
    polarity = TextBlob(review).sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("Review:", review)
    print("Polarity:", round(polarity, 2))
    print("Opinion:", sentiment)
    print()
```
