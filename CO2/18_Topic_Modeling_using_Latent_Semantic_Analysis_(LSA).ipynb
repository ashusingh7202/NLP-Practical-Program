```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

documents = [
    "football players play on the field",
    "football teams score goals",
    "python programming is easy",
    "programming uses python code",
    "players enjoy football games"
]

vectorizer = TfidfVectorizer(stop_words="english")
X = vectorizer.fit_transform(documents)

lsa = TruncatedSVD(n_components=2, random_state=42)
lsa.fit(X)

terms = vectorizer.get_feature_names_out()

for i, topic in enumerate(lsa.components_):
    top_terms = topic.argsort()[-5:][::-1]
    print(f"Topic {i + 1}:",
          [terms[j] for j in top_terms])

print("\nDocument-topic matrix:")
print(lsa.transform(X))
```
