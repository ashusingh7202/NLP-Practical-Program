```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from pathlib import Path

documents = [
    "The football team won the championship",
    "The player scored a goal in the match",
    "Cricket is a popular sport",
    "The coach trained the football players",
    "Artificial intelligence uses machine learning",
    "Python is used for software development",
    "Computer networks connect multiple computers",
    "Deep learning is a branch of AI",
    "Students attend classes in the university",
    "Teachers explain lessons to students",
    "Education improves knowledge and skills",
    "Students prepare for their examinations"
]

labels = [
    "Sports", "Sports", "Sports", "Sports",
    "Technology", "Technology", "Technology", "Technology",
    "Education", "Education", "Education", "Education"
]

X_train, X_test, y_train, y_test = train_test_split(
    documents, labels, test_size=0.25,
    random_state=42, stratify=labels
)

model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("classifier", MultinomialNB())
])
model.fit(X_train, y_train)

predictions = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, predictions))

for doc, predicted in zip(X_test, predictions):
    print("Document:", doc)
    print("Predicted category:", predicted)

new_documents = [
    "The team played a great football match",
    "Python programming builds applications",
    "Students are learning in the classroom",
    "Machine learning is used in artificial intelligence"
]

output_dir = Path("/content/classified_documents")
output_dir.mkdir(exist_ok=True)

for category in ["Sports", "Technology", "Education"]:
    (output_dir / category).mkdir(exist_ok=True)

for i, doc in enumerate(new_documents, start=1):
    category = model.predict([doc])[0]
    path = output_dir / category / f"document_{i}.txt"
    path.write_text(doc, encoding="utf-8")
    print(f"Document {i} saved to {category}")
```
