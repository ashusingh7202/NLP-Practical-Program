```python
from transformers import MarianMTModel, MarianTokenizer

name = "Helsinki-NLP/opus-mt-en-fr"
tokenizer = MarianTokenizer.from_pretrained(name)
model = MarianMTModel.from_pretrained(name)

sentences = [
    "Hello, how are you?",
    "I am learning natural language processing.",
    "Machine learning is very useful."
]

inputs = tokenizer(sentences, return_tensors="pt",
                   padding=True, truncation=True)
translated = model.generate(**inputs)
translations = tokenizer.batch_decode(translated, skip_special_tokens=True)

for original, translation in zip(sentences, translations):
    print("English:", original)
    print("French:", translation)
    print()
```
