```python
from transformers import pipeline

# Using 'text-generation' as 'summarization' is deprecated in transformers v5
generator = pipeline("text-generation", model="gpt2")

text = """
Natural Language Processing is a branch of artificial intelligence
that enables computers to understand and generate human language.
It is used in machine translation, sentiment analysis, chatbots,
text summarization, information extraction, and question answering.
Modern NLP systems use deep learning and transformer architectures.
These models learn language patterns from large collections of text.
They can perform complex tasks, but their output may contain errors.
"""

# Constructing a prompt instructing the model to summarize
prompt = f"Summarize the following text in one sentence:\n{text}\nSummary:"

result = generator(prompt, max_new_tokens=50, do_sample=False)

print("Original text:\n", text)
print("\nSummary:\n", result[0]["generated_text"][len(prompt):].strip())
```
