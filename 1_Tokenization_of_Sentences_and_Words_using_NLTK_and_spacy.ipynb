### Tokenization of Sentences and Words using NLTK



```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
sample_text = "Hello everyone! This is an example of sentence tokenization. We will also tokenize words from one of these sentences."
print(sample_text)
```

```python
from nltk.tokenize import sent_tokenize
sentences = sent_tokenize(sample_text)
print("Sentence Tokenization:")
for i, sentence in enumerate(sentences):
    print(f"Sentence {i+1}: {sentence}")
```

```python
from nltk.tokenize import word_tokenize
first_sentence = sentences[0]
words = word_tokenize(first_sentence)
print(f"\nWord Tokenization of the first sentence ('{first_sentence}'):")
print(words)
```

### Tokenization using spaCy



```python
import spacy

try:
    nlp = spacy.load('en_core_web_sm')
except OSError:
    print('Downloading en_core_web_sm model...')
    !python -m spacy download en_core_web_sm
    nlp = spacy.load('en_core_web_sm')

print("spaCy model loaded successfully!")
```

```python
doc = nlp(sample_text)

# Sentence Tokenization using spaCy
print("\nSentence Tokenization (spaCy):")
spacy_sentences = [sent.text for sent in doc.sents]
for i, sentence in enumerate(spacy_sentences):
    print(f"Sentence {i+1}: {sentence}")
```

```python
if spacy_sentences:
    spacy_first_sentence = nlp(spacy_sentences[0])
    spacy_words = [token.text for token in spacy_first_sentence]
    print(f"\nWord Tokenization of the first sentence ('{spacy_sentences[0]}') (spaCy):")
    print(spacy_words)
else:
    print("No sentences found to perform word tokenization.")
```
