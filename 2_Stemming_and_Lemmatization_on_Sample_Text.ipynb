## Stemming and Lemmatization Example



```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Download necessary NLTK data (if not already downloaded)
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab') # Added to resolve LookupError

print('NLTK libraries and data loaded successfully!')
```

```python
# Sample text
sample_text = "Running, runs, ran, quickly, beautiful, better, best, children, mice, playing, played, plays, studying, studies."

# Tokenize the text
words = word_tokenize(sample_text)
print("Original words:", words)
```

### Stemming



```python
# Initialize Porter Stemmer
porter = PorterStemmer()

# Perform stemming
stemmed_words = [porter.stem(word) for word in words]

print("Stemmed words:", stemmed_words)
```

### Lemmatization



```python
# Initialize WordNet Lemmatizer
lemmatizer = WordNetLemmatizer()


lemmatized_words_noun = [lemmatizer.lemmatize(word, pos='n') for word in words] # Assume noun
lemmatized_words_verb = [lemmatizer.lemmatize(word, pos='v') for word in words] # Assume verb

print("Lemmatized words (assuming noun):", lemmatized_words_noun)
print("Lemmatized words (assuming verb):", lemmatized_words_verb)


print(f"\nLemmatizing 'running' as verb: {lemmatizer.lemmatize('running', pos='v')}")
print(f"Lemmatizing 'running' as noun: {lemmatizer.lemmatize('running', pos='n')}")
```
